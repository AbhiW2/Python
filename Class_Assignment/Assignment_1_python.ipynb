```python
# Variables Example

name = "Abhijeet"
age = 22
college = "SPPU"

print("Name:", name)
print("Age:", age)
print("College:", college)
```

```python
#Basic Data Types

#Integers

age = 22
marks = 95

print(age)
print(marks)
print(type(age))

#Float

cgpa = 8.45
price = 99.99

print(cgpa)
print(price)
print(type(cgpa))


#Boolean
isStudent = True
isPassed = False

print(isStudent)
print(isPassed)
print(type(isStudent))

#String

name = "Abhijeet"
city = "Pune"

print(name)
print(city)
print(type(name))
```

```python
#Arithmetic Operations

a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

```python
#Comparison Operators

a = 15
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
#Assignment Operators

x = 10

x += 5
print(x)

x -= 2
print(x)

x *= 3
print(x)

x /= 2
print(x)

x %= 4
print(x)
```

```python
#String
name = "Python"

print(name)

#String Indexing

text = "Programming"

print(text[0])
print(text[3])
print(text[-1])


#String Slicing
text = "Python Programming"

print(text[0:6])
print(text[7:])
print(text[:6])
print(text[-11:])

#String Operations

first = "Hello"
second = "World"

print(first + " " + second)
print(first.upper())
print(second.lower())
print(len(first))
```

```python
#Lists
fruits = ["Apple", "Banana", "Mango"]

print(fruits)


#Appending to a List
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)

#Inserting into a List
fruits = ["Apple", "Mango"]

fruits.insert(1, "Orange")

print(fruits)


#Removing from a List
fruits = ["Apple", "Mango"]

fruits.insert(1, "Orange")

print(fruits)

fruits.remove("Mango")
print(fruits)


#Sorting a List
numbers = [30, 10, 50, 20]

numbers.sort()

print(numbers)


#Length of a List
fruits = ["Apple", "Banana", "Mango"]

print(len(fruits))


```

```python
# Experiment 1: Swap Two Variables

a = 10
b = 20

print("Before Swapping")
print("a =", a)
print("b =", b)

a, b = b, a

print("\nAfter Swapping")
print("a =", a)
print("b =", b)
```
