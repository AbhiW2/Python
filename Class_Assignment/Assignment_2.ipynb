```python
#create list and accept marks of 5 subjects from user and display average and coresponding grade

list_of_marks = []

for i in range(5):
    mark = float(input(f"Enter marks for subject {i+1}: "))
    list_of_marks.append(mark)

average = sum(list_of_marks) / len(list_of_marks)

if average >= 80:
    grade = "A"
elif average >= 60:
    grade = "B"
else:
    grade = "C"

print(f"Average: {average}")
print(f"Grade: {grade}")
```

```python
#Accept the sentence from user and display the number of words, characters, vowels and consonants in the sentence


sentence = input("Enter a sentence: ")

words = len(sentence.split())
characters = len(sentence)

vowels = 0
consonants = 0

for ch in sentence:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Number of words:", words)
print("Number of characters:", characters)
print("Number of vowels:", vowels)
print("Number of consonants:", consonants)
```
