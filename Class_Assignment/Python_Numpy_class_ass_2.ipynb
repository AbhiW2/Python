```python
import numpy as np

def temperature_analysis(temperatures):

    average = np.mean(temperatures)

    maximum = np.max(temperatures)

    minimum = np.min(temperatures)

    highest_day = np.argmax(temperatures) + 1

    above_average = temperatures[temperatures > average]

    return average, maximum, minimum, highest_day, above_average


temperatures = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])

result = temperature_analysis(temperatures)

print("Average temperature:", result[0])
print("Maximum temperature:", result[1])
print("Minimum temperature:", result[2])
print("Day with highest temperature: Day", result[3])
print("Temperatures above average:", result[4])
```

```python
import numpy as np

marks = np.array([
    [78, 82, 91],  
    [65, 74, 80],  
    [92, 88, 95],  
    [56, 69, 72],  
    [84, 90, 87]   
])

student_avg = np.mean(marks, axis=1)

subject_avg = np.mean(marks, axis=0)

highest = np.max(marks)

lowest = np.min(marks)

highest_student = np.argmax(student_avg) + 1

subjects = ["Mathematics", "Physics", "Programming"]
highest_subject = subjects[np.argmax(subject_avg)]

print("Marks Matrix:")
print(marks)

print("\nAverage marks of each student:")
for i, avg in enumerate(student_avg):
    print(f"Student {i+1}: {avg:.2f}")

print("\nAverage marks in each subject:")
for subject, avg in zip(subjects, subject_avg):
    print(f"{subject}: {avg:.2f}")

print("\nHighest mark:", highest)
print("Lowest mark:", lowest)
print("Student with highest average: Student", highest_student)
print("Subject with highest average:", highest_subject)
```

```python
import numpy as np

def transform_data(data):
    data = np.array(data)

    square = data ** 2
    square_root = np.sqrt(data)
    exponential = np.exp(data)

    mean_square = np.mean(square)
    mean_square_root = np.mean(square_root)
    mean_exponential = np.mean(exponential)

    print("Original Data:", data)
    print("Square:", square)
    print("Square Root:", square_root)
    print("Exponential:", exponential)

    print("Mean of Square:", mean_square)
    print("Mean of Square Root:", mean_square_root)
    print("Mean of Exponential:", mean_exponential)


measurements = [2, 4, 6, 8, 10, 12]

transform_data(measurements)
```

```python
import numpy as np

experiment_A = np.array([12, 15, 14, 18, 20, 17, 16])
experiment_B = np.array([10, 13, 15, 16, 19, 21, 18])

# Mean
mean_A = np.mean(experiment_A)
mean_B = np.mean(experiment_B)

std_A = np.std(experiment_A)
std_B = np.std(experiment_B)

max_A = np.max(experiment_A)
max_B = np.max(experiment_B)

min_A = np.min(experiment_A)
min_B = np.min(experiment_B)

difference = experiment_A - experiment_B

print("Experiment A")
print("Mean:", mean_A)
print("Standard Deviation:", std_A)
print("Maximum:", max_A)
print("Minimum:", min_A)

print("\nExperiment B")
print("Mean:", mean_B)
print("Standard Deviation:", std_B)
print("Maximum:", max_B)
print("Minimum:", min_B)

print("\nDifference between experiments:")
print(difference)

if std_A > std_B:
    print("\nExperiment A shows greater variation.")
else:
    print("\nExperiment B shows greater variation.")
```

```python
import numpy as np


# Function 1: Create random dataset
def create_dataset():
    data = np.random.randint(1, 101, size=(10, 5))
    return data


# Function 2: Analyze the dataset
def analyze_data(data):

    print("Dataset:")
    print(data)

    # Shape and dimensions
    print("\nShape:", data.shape)
    print("Dimensions:", data.ndim)

    # Mean of each variable
    variable_mean = np.mean(data, axis=0)
    print("\nMean of each variable:")
    print(variable_mean)

    # Mean of each experiment
    experiment_mean = np.mean(data, axis=1)
    print("\nMean of each experiment:")
    print(experiment_mean)

    # Maximum and minimum
    print("\nMaximum value:", np.max(data))
    print("Minimum value:", np.min(data))

    # Overall mean
    overall_mean = np.mean(data)
    print("Overall Mean:", overall_mean)

    # Values above overall mean
    above_mean = data[data > overall_mean]
    print("\nValues above overall mean:")
    print(above_mean)

    # Reshape the data
    reshaped_data = data.reshape(5, 10)
    print("\nReshaped Data (5 x 10):")
    print(reshaped_data)

    # Flatten the reshaped data
    flattened_data = reshaped_data.flatten()
    print("\nFlattened Data:")
    print(flattened_data)


# Create dataset
dataset = create_dataset()

# Analyze dataset
analyze_data(dataset)
```

```python
import numpy as np


def generate_marks():
    marks = np.random.randint(0, 101, size=(10, 4))
    return marks


def get_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


def analyze_students(marks):

    student_averages = np.mean(marks, axis=1)

    subject_averages = np.mean(marks, axis=0)

    highest_student = np.argmax(student_averages)
    lowest_student = np.argmin(student_averages)

    highest_mark = np.max(marks)
    lowest_mark = np.min(marks)

    students_above_75 = np.sum(student_averages > 75)

    marks_above_90 = marks[marks > 90]

    student_information = []

    for i in range(10):
        grade = get_grade(student_averages[i])

        student = {
            "Student": i + 1,
            "Marks": marks[i],
            "Average": student_averages[i],
            "Grade": grade
        }

        student_information.append(student)

    return (student_averages, subject_averages,
            highest_student, lowest_student,
            highest_mark, lowest_mark,
            students_above_75, marks_above_90,
            student_information)



marks = generate_marks()

result = analyze_students(marks)

student_averages = result[0]
subject_averages = result[1]
highest_student = result[2]
lowest_student = result[3]
highest_mark = result[4]
lowest_mark = result[5]
students_above_75 = result[6]
marks_above_90 = result[7]
student_information = result[8]


print("=" * 50)
print("       STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

print("\nMarks of Students:")
print(marks)

print("\nAverage Mark of Each Student:")
for i in range(10):
    print(f"Student {i + 1}: {student_averages[i]:.2f}")


print("\nAverage Mark of Each Subject:")
for i in range(4):
    print(f"Subject {i + 1}: {subject_averages[i]:.2f}")


print("\nHighest Average:")
print(f"Student {highest_student + 1} "
      f"with average {student_averages[highest_student]:.2f}")

print("\nLowest Average:")
print(f"Student {lowest_student + 1} "
      f"with average {student_averages[lowest_student]:.2f}")


print("\nHighest Mark in Dataset:", highest_mark)
print("Lowest Mark in Dataset:", lowest_mark)

print("\nNumber of students with average above 75:",
      students_above_75)

print("\nAll marks greater than 90:")
print(marks_above_90)


print("\nFinal Student Information:")
print("-" * 50)

for student in student_information:
    print(
        f"Student {student['Student']} | "
        f"Marks: {student['Marks']} | "
        f"Average: {student['Average']:.2f} | "
        f"Grade: {student['Grade']}"
    )

print("=" * 50)
```
