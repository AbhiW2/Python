```python
#reverse a given number

num=int(input("Enter a number: "))

rev=0

while(num>0):
    dig=num%10
    rev=rev*10+dig
    num=num//10

print("Reverse of the given number is:", rev)
```

```python
#Temperature converter

print("Temperature converter:")
print("1.celsius to fahrenhiet:")
print("fahrenheit to celsius:")

choice=int(input("Enter your choice from 1 to 2:"))

temp=int(input("Enter temperature:"))

if choice == 1:
    fahrenheit=(temp*9/5)+32
    print("Temperature in ferinite:",fahrenheit)
elif choice ==2:
    celsius=(temp-32)*5/9
    print("Temperature in celcius:",celsius)
else:
    print("Invalid choice please enter valid choice between 1 to 2.....")
```

```python
#Calculate Area and perimeter
import math
pi=3.14;

#For circle
radius=float(input("\nEnter radius of circle:"))

circle_area=pi*radius*radius;
circle_perimeter=2*pi*radius;

print("Area of circle:",circle_area)
print("Perimeter of circle:",circle_perimeter)

#for Rectangle
length=int(input("\nEnter length of rectangle:"))
width=int(input("\nEnter Width of rectangle:"))

rect_length=length*width;   
rect_width=2*(length+width)

print("Length of rectangle:",rect_length)
print("Width of Rectangle:",rect_width)


#for triangle
a = float(input("\nEnter first side of triangle: "))
b = float(input("Enter second side of triangle: "))
c = float(input("Enter third side of triangle: "))

s = (a + b + c) / 2

triangle_area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("\nTriangle Area:", triangle_area)
```

```python

principal = float(input("Enter principal amount: "))
rate = float(input("Enter annual interest rate (%): "))
years = int(input("Enter number of years: "))

# Calculate simple interest
interest = (principal * rate * years) / 100

# Calculate total amount
total_amount = principal + interest

print("\nSimple Interest:", interest)
print("Total Amount:", total_amount)

# Amount at the end of each year
print("\nAmount at the end of each year:")

for year in range(1, years + 1):
    amount = principal + (principal * rate * year) / 100
    print("Year", year, ":", amount)
```
