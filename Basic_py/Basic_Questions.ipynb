```python
#Check Given num even or odd

num=int(input("Enter a number: "))
print("Given number is: ",num)

if(num%2==0):
    print("Your Entered number is Even:", num)
else:
    print("Your Entered number is Odd:", num)
```

```python
#Find Greatest of three numbers entered by user

a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
c=int(input("Enter third number: "))

if(a>b and a>c):
    print("First number is greatest:", a)
elif(b>a and b>c):
    print("Second number is greatest:", b)
else:
    print("Third number is greatest:", c)   
```
