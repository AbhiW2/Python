```python
#creating numpy array

import numpy as np

num=np.array([2,3,4,5,6,7,8,9])
print(num)
```

```python
#Dimension of Numpy Array

a=np.array([1,2,3,4,5]) #one dimension

print(a)

b=np.array([
    [1,2,3,], #Two dimension
    [5,6,7]

])
print(b)



c = np.array([
    [
        [1, 2],  #3 dimesion
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print(c)
```

```python
#Check how many dimension are present in array used 
#nidm

print(a.ndim)
print(b.ndim)
print(c.ndim)
```

```python
#Check shape of array along each dimensions
#used shape

print(a.shape)
print(b.shape)
```

```python
#Check size of give array how many num are present in array
#used size

print(a.size)
print(b.size)
print(c.size)

#check which data type of array
print(a.dtype)

```

```python
#Indexing the array 
#indexing means accing a particular array element

d=np.array([1,2,3,4,5,6,7])
print(d[4])
```

```python
#Slicing 
#slicing means taking a portion of an array

print(d[1:4])
```

```python
#Mathematical operators
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
#Important Numpy functions

#np.zeros()
#Creates an array filled with zeros.
a = np.zeros(5)
print(a)

#np.ones
a = np.ones(5)
print(a)

#np.arange
#creates values within range
a = np.arange(1, 10)
print(a)


#np.linspace()
#Creates evenly spaced numbers.
a = np.linspace(0, 10, 5)
print(a)


#reshape
#reshape() changes the structure/shape of an array without changing its data.
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print(b)



#Aggregation functions

a = np.array([10, 20, 30, 40])
print(np.sum(a)) #Calculate sum of array

print(np.mean(a))#Calculate mean

print(np.max(a))#calculate maximum in array

print(np.min(a))


print(np.std(a))#calculate standard deviation in array


```
