```python
#Series 
#A Series is basically one column of data.

import pandas as pd

marks = pd.Series([80, 90, 75, 88])

print(marks)
```

```python
#DataFrame
#A DataFrame is a table containing rows and columns.

data = {
    "Name": ["Amit", "Rahul", "Priya"],
    "Age": [21, 22, 20],
    "Marks": [85, 72, 91]
}

df = pd.DataFrame(data)

print(df)

#head
#Shows the first rows.
print(df.head())

#tail
```

```python
#Info
#provied information about data frame
df.info()
```

```python
#describe
#It provides statistical information of df
print(df.describe())
```

```python
print(df.shape)
```

```python
print(df.columns)
print(df["Name"])
print(df["Marks"])
print(df[["Name", "Marks"]])
```

```python
#loc[]
#Used to select data using labels/names.
print(df.loc[0])
print(df.loc[0, "Name"])
print(df.loc[df["Marks"] > 80])
```

```python
#iloc[]
#iloc means integer-location based indexing.
print(df.iloc[0])
```
