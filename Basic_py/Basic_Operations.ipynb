```python
#Relational operations
a=10
b=23
print("a is equal to b:",a==b)
print("a is not equal to b:",a!=b)
print("a is greater than b:",a>b)
print("a is less than b:",a<b)
print("a is greater than or equal to b:",a>=b)
print("a is less than or equal to b:",a<=b)
```

```python
#Python Data Structures Operations

#List Operations

list1=[1,2,3,4,5]
list2=[6,7,8,9,10]

print("List1 is:",list1)
print("List2 is:",list2)

print("Concatenation of list1 and list2 is:",list1+list2)

list1.append(11)
print("List1 after appending 11 is:",list1)

list1.insert(0,11)
print("List1 after inserting 11 at index 0 is:",list1)
list1.remove(3)
print("List1 after removing 3 is:",list1)
list1.pop(4)
print("List1 after popping element at index 4 is:",list1)

list2.sort()
print("List2 after sorting is:",list2)
list2.reverse()
print("List2 after reversing is:",list2)


```

```python
#Tuple Operations

tuple=(2,3,4,5,6,7,8,9,10)
print("Tuple is:",tuple)
print("Element at index 2 is:",tuple[2])
print("Slice of tuple from index 0 to 5 is:",tuple[0:5])


```
